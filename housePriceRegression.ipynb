```python
CategoricC = df.select_dtypes(include=["str","object"]).columns
CategoricC
```

```python
New = df.columns.drop(['Id', 'SalePrice'])
```

```python
NoNeed = ["Id"]
for i in New:
    MostFrequently = df[i].value_counts(normalize=True, dropna=False).iloc[0]
    if MostFrequently > 0.95:
        NoNeed.append(i)
        print(f"{i:15s} most frequently value percent : {MostFrequently:.3f}")

NumericC = [c for c in NumericC if c in df.columns]
CategoricC = [c for c in CategoricC if c in df.columns]
df = df.drop(columns=NoNeed)
print("\nDeleted",NoNeed)
```

```python
NumericC = [c for c in NumericC if c in df.columns]
CategoricC = [c for c in CategoricC if c in df.columns]
```

```python
for i in CategoricC:
    df[i] = df[i].fillna("None")

df['LotFrontage'] = df.groupby('Neighborhood')['LotFrontage'].transform(lambda x: x.fillna(x.median()))
df['MasVnrArea'] = df['MasVnrArea'].fillna(0)
df['GarageYrBlt'] = df["GarageYrBlt"].fillna(0)

for i in NumericC:
    if df[i].isnull().any():
        df[i] = df[i].fillna(df[i].median())

print('StillNuls:', df.isnull().sum().sum())
```

```python
EncodingDf = pd.get_dummies(df, columns=CategoricC, drop_first=True, dtype=int)
EncodingDf.head()
```

```python
X = EncodingDf.drop(columns='SalePrice')
y = EncodingDf['SalePrice']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=28)
print(X_train.shape, X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaler = scaler.fit_transform(X_train)
X_test_scaler = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
sonuclar = []

def degerlendir(isim, model, XTS=X_test_scaler):
    pred = model.predict(XTS)
    mae = mean_absolute_error(y_test, pred)
    mse = mean_squared_error(y_test, pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, pred)
    sonuclar.append({'Model': isim, 'MAE': mae, 'MSE': mse, 'RMSE': rmse, 'R2': r2})
    print(f'{isim}\n  MAE {mae:,.0f} | MSE {mse:,.0f} | RMSE {rmse:,.0f} | R2 {r2:.4f}')
```

```python
LinReg = LinearRegression().fit(X_train_scaler, y_train)
degerlendir('Linear Regression', LinReg)
```

```python
ridge = Ridge(alpha=10).fit(X_train_scaler, y_train)
degerlendir('Ridge (alpha=10)', ridge)
```

```python
lasso = Lasso(alpha=10, max_iter=10000).fit(X_train_scaler, y_train)
degerlendir('Lasso (alpha=10)', lasso)
print('Lasso sıfırladığı katsayı sayısı:', (lasso.coef_ == 0).sum(), '/', len(lasso.coef_))
```

```python
ElNet = ElasticNet(alpha=0.3, l1_ratio=0.5, max_iter=10000).fit(X_train_scaler, y_train)
degerlendir('ElasticNet (alpha=0.3, l1=0.5)', ElNet)
```

```python
from sklearn.model_selection import cross_val_score, KFold
```

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)
models = {'Linear': LinReg, 'Ridge': ridge, 'Lasso': lasso, 'ElasticNet': ElNet}
Sonuc = {}

for isim, m in models.items():
    skor = cross_val_score(m, X_train_scaler, y_train, cv=kf, scoring='r2')
    Sonuc[isim] = skor
    print(f'{isim:12s} CV R2: {skor.mean():.4f} ({skor.std():.4f})')
print(Sonuc)
```

```python
plt.figure(figsize=(8, 4))
sns.boxplot(data=pd.DataFrame(Sonuc))
plt.ylabel('CV R2'); plt.title('5-Fold CV karşılaştırma'); plt.show()
```

```python
from sklearn.model_selection import GridSearchCV

RidgeG = GridSearchCV(Ridge(), {'alpha': np.logspace(-1, 3, 30)},cv=kf, scoring='r2', n_jobs=-1).fit(X_train_scaler, y_train)

LassoG = GridSearchCV(Lasso(max_iter=20000), {'alpha': np.logspace(0,3, 15)},cv=kf, scoring='r2', n_jobs=-1).fit(X_train_scaler, y_train)

ElNetG = GridSearchCV(ElasticNet(max_iter=20000), {'alpha': np.logspace(0,3, 15),
'l1_ratio': [0.1, 0.3, 0.5, 0.7, 0.9, 0.95, 1.0]}, cv=kf, scoring='r2', n_jobs=-1).fit(X_train_scaler, y_train)

print('Ridge  best alpha:', RidgeG.best_params_, '| CV R2:', RidgeG.best_score_)
print('Lasso  best alpha:', LassoG.best_params_, '| CV R2:', LassoG.best_score_)
print('ENet best param:', ElNetG.best_params_, '| CV R2:', ElNetG.best_score_)
```

```python
degerlendir("Ridge", RidgeG.best_estimator_)
degerlendir("Lasso", LassoG.best_estimator_)
degerlendir("ElasticNet", ElNetG.best_estimator_)
```

```python
BestScore = LassoG.best_estimator_

coefficients = pd.Series(BestScore.coef_, index=X.columns)

TopIndex = coefficients.abs().nlargest(20).index
Top = coefficients.reindex(TopIndex)

plt.figure(figsize=(8, 7))
sns.barplot(x=Top.values, y=Top.index, color='turquoise')
plt.title('Lasso - Top 20 Most Influential Features');
plt.show()
```

```python
pred = BestScore.predict(X_test_scaler)

plt.figure(figsize=(6, 6))
plt.scatter(y_test, pred, alpha=0.5)
plt.plot([0, 800000], [0, 800000], 'r--')
plt.xlabel('Actual Price')
plt.ylabel('Predicted Price')
plt.title('Actual vs Predicted (Lasso tuned)')
plt.show()
```
